```python
import torch
import json
from torchvision import datasets
import torch.optim as optim
from helper import get_preprocessing_pipeline, test_processor, train_processor
from torch.optim.lr_scheduler import StepLR
from the_net import Net
```

## Setup

```python
EPOCHS = 14



GAMMA = 0.7
LR = 1.0
```

```python
device = torch.device("cpu")
transform = get_preprocessing_pipeline()
dataset1 = datasets.MNIST('../data', train=True, download=True, transform=transform)
dataset2 = datasets.MNIST('../data', train=False, transform=transform)
```

```python
train_loader = torch.utils.data.DataLoader(dataset1, batch_size=64, shuffle=True)
test_loader = torch.utils.data.DataLoader(dataset2, batch_size=1000, shuffle=False)
```

```python
model = Net().to(device)
optimizer = optim.Adadelta(model.parameters(), lr=LR)

scheduler = StepLR(optimizer, step_size=1, gamma=GAMMA)
for epoch in range(1, EPOCHS + 1):
    train_processor(model, device, train_loader, optimizer, epoch)
    test_processor(model, device, test_loader)
    scheduler.step()
```

## Saving the Model

```python
torch.save(model.state_dict(), "mnist_cnn.pt")
```
